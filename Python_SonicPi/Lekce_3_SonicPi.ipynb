## Kapitola 3.2: Logické výrazy

V programování, stejně jako ve výrokové logice, pracujeme s tvrzeními, která jsou buď pravdivá (True), nebo nepravdivá (False). Pomocí logických spojek (operátorů) můžeme z těchto jednoduchých výroků sestavovat komplexní formule, které určují chování našeho programu.

Základní logické operátory v Pythonu odpovídají standardním logickým funkcím:

  * and (Konjunkce ∧): Celý výraz je pravdivý pouze tehdy, jsou-li všechny jeho části (podvýrazy) pravdivé. Stačí jedna nepravda a celý výsledek je False.
  * or (Disjunkce ∨): Výraz je pravdivý, pokud je alespoň jeden z jeho podvýrazů pravdivý. K dosažení výsledku False musí být nepravdivé úplně všechny části.
  * not (Negace ¬): Jedná se o unární operátor, který obrací pravdivostní hodnotu – z pravdy dělá nepravdu a naopak.

Ve světě výrokové logiky existují i další důležité operace. V Pythonu pro ně sice nemáme samostatná klíčová slova, ale můžeme je snadno sestavit kombinací těch základních:

  * Implikace (⇒): Vztah „jestliže A, pak B“. V Pythonu ji zapíšeme jako (not A) or B.
  * Ekvivalence / Totožnost (⇔): Výraz je pravdivý, pokud mají oba podvýrazy stejnou hodnotu. Zapíšeme jako A == B.
  * XOR / Různoznačnost (⊕): Pravda nastává právě tehdy, když se hodnoty A a B liší. Zapíšeme jako A != B.
  * Inhibice: Situace, kdy platí A, ale zároveň neplatí B. Zapíšeme jako A and not B.

V buňce níže se zahraje vítězný tón, pouze pokud budou splněny všechny dílčí výroky najednou:

Kromě základního else (všechno ostatní) můžeme využít i konstrukci elif (zkratka pro else if). Ta nám umožní přidat do programu libovolné množství dalších specifických podmínek.<br>Příkaz elif nemůže existovat samostatně – musí vždy následovat po úvodním if a předcházet případnému závěrečnému else.
```python
hudebni_styl = input("Zadej hudební styl (techno/ambient/pop/rock): ")

if hudebni_styl == "techno":
    sample(DRUM_HEAVY_KICK)
    sleep(0.5)
    sample(DRUM_HEAVY_KICK)
    sleep(0.5)
    sample(DRUM_HEAVY_KICK)

elif hudebni_styl == "ambient":
    use_synth(HOLLOW)
    play(60, attack=2, release=4)
    play(67, attack=2, release=4)

elif hudebni_styl == "pop":
    use_synth(PRETTY_BELL)
    play(72)
    sleep(0.2)
    play(76)
    sleep(0.2)
    play(79)

elif hudebni_styl == "rock":
    use_synth(SAW)
    play(40, release=1, cutoff=100)
    sleep(0.5)
    play(40, release=0.5)

else:
    use_synth(PIANO)
    play(50, release=0.1)
    sleep(0.1)
    play(50, release=0.1)
```

Vykonání úlohy výše se řídí tedy podle těchto bodů:
1) První kontrola (if): Program se podívá na první podmínku. Pokud je pravdivá, zahraje příslušný zvuk a zbytek celého bloku (všechny elif i else) okamžitě přeskočí.

2) Hledání shody (elif): Pokud první podmínka neplatí, program se přesune k prvnímu elif a zkoumá jeho pravdivost. Takto postupně „testuje“ jeden elif za druhým, dokud nenarazí na ten, který odpovídá zadanému stylu.

3) Konečné řešení (else): Pokud program projde všechny podmínky if i elif a ani jedna z nich „nezazněla“ (nebyla pravdivá), automaticky vykoná instrukce v bloku else.

```python
nastroj = input("Vyber nástroj (bici / klavir): ")
nalada = input("Vyber náladu (vesela / temna): ")

if nastroj == "bici":
    if nalada == "vesela":
        sample(DRUM_CYMBAL_SOFT)
    elif nalada == "temna":
        sample(DRUM_HEAVY_KICK)
    else:
        play(50)

elif nastroj == "klavir":
    use_synth(PIANO)
    if nalada == "vesela":
        play(72)
        sleep(0.1)
        play(76)
    elif nalada == "temna":
        play(36, release=3)
    else:
        play(50)

else:
    use_synth(SAW)
    play(40, release=0.1)
```
