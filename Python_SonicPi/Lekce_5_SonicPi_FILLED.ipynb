## Kapitola 5.2: Vyhledávání hodnot

Klíčovým tématem je zjištění přítomnosti a pozice prvku v kolekci pomocí algoritmů.<br>
Následující program hledá ton 50:

  * Nalezeno: Program zahraje potvrzení přehráním nalezeného tónu a pomocí _break_ předčasně ukončí cyklus pro úsporu zdrojů.
  * Nenalezeno: Pokud cyklus proběhne celý bez přerušení, provede se blok else přehráním:<br>
                use_synth(SAW)<br>
                play(37, release=0.2)<br>
                play(38, release=0.2)<br>
                který signalizuje, že prvek v seznamu není.

````python
tony = [31, 34, 37, 39, 42, 43, 45, 48, 50, 52, 53, 55, 58, 60, 61, 63, 64, 66, 67, 69]

for cislo in tony:
    if cislo == 50:
        use_synth(BEEP)
        play(cislo)
        break
else:
    use_synth(SAW)
    play(37, release=0.2)
    play(38, release=0.2)
````

Hledání extrémů (minima a maxima) je další nejen programátorské cvičení. Je to základní kámen analýzy dat, se kterým se setkáváte v aplikacích každý den. Algoritmus pro hledání maxima a a minima je náledující:
* Označ prvek na indexu 0 za nejnižší/nejvyšší nalezené číslo.
* V cyklu projdi celou kolekci od prvního prvku po konec a u každé prvku se zeptej, zda je vetší/menší, než naše dosavadní maximum/minimum.
* Pokud ano, přepíšeme hodnotu v proměnné maximum/minimum a jdeme k dalšímu prvku, dokud neprojdeme celou kolekci(seznam).<br>
Algoritmus pro heldání maxima:

na závěr si proměnnou s maximální hodnotou přehrajeme

````python
tony = [53, 34, 61, 45, 67, 31, 58, 42, 39, 48, 55, 60, 37, 66, 43, 63, 52]

max_cislo = tony[0]
for idx in range(1, len(tony)):
    if tony[idx] > max_cislo:
        max_cislo = tony[idx]
play(max_cislo)
````

#### Cvičení 5.2.3: Vyhledání indexu maxima
Upravte předchozí kód tak, aby vrátil i index, na kterém se nachází maximální hodnota.
opět použijme v příkazu play paramter pan = (index_hledane_noty / len(melodie)) * 2 - 1 pro určení indexu maxima

````python
tony = [53, 34, 61, 45, 67, 31, 58, 42, 39, 48, 55, 60, 37, 66, 43, 63, 52]

max_cislo = tony[0]
for idx in range(1, len(tony)):
    if tony[idx] > max_cislo:
        max_cislo = tony[idx]
        max_index = idx
play(max_cislo, pan = (max_index / len(tony)) * 2 - 1)
````

#### Cvičení 5.2.4: Vyhledání minima
Napište kód, který v kolekci nalezne nejmenší číslo.

````python
tony = [53, 34, 61, 45, 67, 31, 58, 42, 39, 48, 55, 60, 37, 66, 43, 63, 52]

min_cislo = tony[0]
for idx in range(1, len(tony)):
    if tony[idx] < min_cislo:
        min_cislo = tony[idx]
        min_index = idx
play(min_cislo, pan = (min_index / len(tony)) * 2 - 1)
````

#### Cvičení 5.3.4: Prohazování dvojic
Napište program, který projde všechny dvojice stejným způsobem jako v předchozím cvičení, ale pokud je prvek vlevo větší jak prvek vpravo, tak se mezi sebou prohodí.

```
30 < 34 neprohodíme -> [30,34,32,36,38,44,46,40,42,48,50]
34 > 32 prohodíme -> [30,32,34,36,38,44,46,40,42,48,50]
34 < 36 neprohodíme -> [30,32,34,36,38,44,46,40,42,48,50]
a tak dále...
```

Přehrajte postupně tóny v kolekci před a po prohazování pro porovnání správnosti řešení.


````python
tony = [30, 34, 32, 36, 38, 44, 46, 40, 42, 48, 50]

#Přehrání kolekce tony před řazením
for i in range(len(tony)):
  play(tony[i])
  sleep(0.5)

for i in range(len(tony)-1):
    if tony[i] > tony[i+1]:
        tony[i], tony[i+1] = tony[i+1], tony[i]

sleep(1)
#Přehrání kolekce tony po jednom kole řazení
for i in range(len(tony)):
  play(tony[i])
  sleep(0.5)
````
