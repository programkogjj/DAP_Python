#### Kapitola 6.1 Definice uživatelem-definované funkce
Uživatelem definovaná funkce začíná klíčovým slovem def, za kterým následuje Vámi zvolený název funkce. Poté následuje v závorkách libovolný počet parametrů. Parametr je hodnota, se kterou funkce uvnitř může pracovat.
```        
def nazev_funkce(parametr1, parametr2, ..., parametrN):
    ...
    return navratova_hodnota
```
Funkce je pak volána z hlavní části programu s doplněnými hodnotami pro parametry. Těmto hodnotám se říká argumenty. Návratovou hodnotu z funkce můžeme uložit do proměnné, se kterou v hlavní části kódu dále pracujeme.

V následující buňce vidíte funkci s názvem ```secti_hodnoty()```, která má 2 parametry, cislo ```a``` a číslo ```b```. Při zavolání funkcem musíme do těchto 2 parametrů předat skutečnou hodnotu, čili napsat 2 agrumenty. konkrétně předávám hodnoty 30 a 35.<br> Návratová hodnota v proměnné součet uvnitř funkce bude uložena do proměnné ```soucet``` v hlavní části kódu. Argumenty se postupně dosazují za parametry funkce, takže pořadí je důležité. Argumenty zavedené tímto způsobem do funkce nazýváme poziční argumenty (args).

````python
def secti_hodnoty(a, b):
  soucet = a + b
  return soucet

vysledek = secti_hodnoty(30, 35)
play(vysledek)
````

Při volání funkcí lze využít explicitní přiřazování hodnot k jednotlivým parametrům pomocí zápisu parametr=argument, což odstraňuje závislost na jejich definovaném pořadí. Tento přístup, označovaný jako keyword arguments (kwargs), umožňuje dosazovat argumenty v libovolném pořadí (např. nejprve pro parametr b a následně pro a), neboť každá hodnota je jednoznačně svázána se svým klíčovým slovem, jak ilustruje praktická ukázka v buňce níže.

Tento způsob vkládání lze i kombinovat. Část parametrů lze doplnit jen uvedením hodnoty a část skrze parametry. Je však důležité dodržet následující pravidlo: nejprve musíme vyplnit všechny args a poté až kwargs.

Kromě typů můžete funkci popsat i slovně pomocí docstringu. Píše se do trojitých uvozovek (""") hned pod definici funkce. Technicky jde o víceřádkový řetězec, který Python interpretuje jako dokumentaci.

#### Cvičení 6.2.1: Dokumentace hudební funkce

Vytvořme jednoduchou funkci vypocitej_dozvuk.
  *  Přidejme napovídání typů pro všechny parametry i návratovou hodnotu.
  *  Doplňme dokumentační řetězec (docstring), který vysvětlí, co funkce dělá.
  *  Parametry jsou: nota a síla.
  *  Hodnotu dozvuku spočteme jako nota / síla a uložíme do proměnné výsledek
  *  Funkce vrací vypočtený výsledek jako desetinné číslo (float).

Funkci otestujte

Kromě základních datových typů, jako je:

    celé číslo (int)
    desetinné číslo (float)
    řetězec (str)
    pravdivostní hodnota (bool)

potřebujete dokumentovat i kolekce hodnot. Například seznam celých čísel. Pro takové dokumentování musíme využít pokročilejší techniku a to jsou dokumentační datové struktury z modulu typing. V následujícím příkladu je dokumentováno, že do funkce vstupuje seznam celých čísel.<br>
Termín None na konci prvního řádku říká, že funkce nemá žádnou návratovou hodnotu -> V některých zdrojích se takovým funcím říká procedura

Níže vidíte kód, který už jsme měli v 5. lekci. Program nalezne maximální hodnotu v seznamu a přehraje jí. Pojďme si do ní přidat docstringy a datové typování

````python
def prehraj_max(cisla: list[int]) -> None:
  """
  Zjistí maximální hodnotu v zadaném seznamu čísel a přehraje ji

  Args:
    cisla (list[int]): seznam s číselnými hodnotami.

  Navratová hodnota: None
  """
  max_cislo = cisla[0]
  for idx in range(1, len(cisla)):
      if cisla[idx] > max_cislo:
          max_cislo = cisla[idx]
  play(max_cislo)
````

## Kapitola 6.3: Strukturované programování

Funkce si můžeme představit jako izolované světy. Z okolního prostředí přijmou vstup (argumenty dosazené do parametrů), zpracují je podle svého algoritmu a výsledek pošlou zpět tam, odkud byly zavolány.

Tento přístup nám umožňuje zásadní změnu v uvažování: náš hlavní program se stává "manažerem", který volá funkce jako specializované služby. Funkce vykonají konkrétní úkol a předají výsledek dál. Hlavní program tyto výsledky sbírá a může je řetězit – výstup jedné funkce se stane vstupem pro druhou. Funkce navíc mohou volat i samy sebe navzájem, což nám dává obrovskou svobodu v organizaci kódu.

V profesionálním programování je zvykem psát hlavní logiku programu do funkce s názvem main() (anglicky „hlavní“). Tím se kód zpřehlední a oddělí se „příprava“ od „samotného spuštění“.

Protože je ale main() také jen funkce, musíme ji na konci programu ručně zavolat, aby se celý proces spustil.

````python
import random

def generuj_motiv() -> list[int]:
    akord = []
    for i in range(4):
      ton = random.randint(40,80)
      akord.append(ton)
    return akord

def analyzuj_energii(noty: list[int]) -> float:
    if not noty:
        return 0.0
    return sum(noty) / len(noty) / 100

def zahraj_vysledek(noty: list[int], sila: float) -> None:
    use_synth(PIANO)
    for nota in noty:
        play(nota, amp=sila, release=0.8)
        sleep(0.5)


def main():
    moje_noty = generuj_motiv()
    vypoctena_hlasitost = analyzuj_energii(moje_noty)
    zahraj_vysledek(noty=moje_noty, sila=vypoctena_hlasitost)


main()
````

V praxi můžete při pohledu na cizí kód na internetu často narazit na toto volání schované v podmínce

````python
if __name__ == "__main__":
    main()
````

Pro naše účely v Sonic Pi stačí vědět, že dělá úplně to samé jako prosté main() –> spustí váš hlavní program. Není třeba se ho bát, jen ho berte jako „formálnější pozdrav“ na konci kódu. Pokud bychom začali náš program rozdělovat do více souborů, pak je tento přístup nutný.

````python
import random

def generuj_motiv() -> list[int]:
    akord = []
    for i in range(4):
      ton = random.randint(40,80)
      akord.append(ton)
    return akord

def analyzuj_energii(noty: list[int]) -> float:
    if not noty:
        return 0.0
    return sum(noty) / len(noty) / 100

def zahraj_vysledek(noty: list[int], sila: float) -> None:
    use_synth(PIANO)
    for nota in noty:
        play(nota, amp=sila, release=0.8)
        sleep(0.5)

def main():
    moje_noty = generuj_motiv()
    vypoctena_hlasitost = analyzuj_energii(moje_noty)
    zahraj_vysledek(noty=moje_noty, sila=vypoctena_hlasitost)


if __name__ == "__main__":
    main()
````
