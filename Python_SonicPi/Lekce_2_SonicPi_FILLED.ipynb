## Kapitola 2.3: Rozdělení řetězce

Z výzvy jsme schopni přijímat pouze jeden řetězec. To může být nepohodlné v situacích, kdybychom chtěli zadat naráz více informací. Takový případ typicky nastává u zadávání více čísel, které můžeme oddělit ve výzve mezerou, čárkou nebo jiným separátorem (znak pro oddělení informací).

Řešení situace spočívá v rozdělení řetězce příkazem _split()_, který má jako parametr řetězec, představující separátor. Příkaz následně navrátí kolekci hodnot, konkrétně kolekci s názvem _kolekce_. V rámci lekce 2 musíme ještě udělat jeden krok, a to "rozbalit" hodnoty z kolekce do jednotlivých proměnných.<br>
Hrát jednotlivé tóny přímo z kolekce se naučíme v budoucích lekcích

Hodnoty uložené z kolekce můžeme uložit do správného počtu proměnných pomocí tzv. rozbalování (unpacking). Na následující buňce vidíte rozbalení seznam tří celých čísel do tří proměnných a,b,c. Tyto proměnné musí být na levé straně rovnice odděleny čárkou (mezery jsou dobrovolné) a jejich počet musí odpovídat počtu hodnot v kolekci na pravé straně.


<div style="color:#90EE90">

#### Tip pro vyučující:

Při rozbalování (např. A, B, C = seznam) musí počet proměnných vlevo přesně odpovídat počtu prvků v seznamu vpravo. Pokud student zadá 4 čísla místo 3, program spadne s ValueError.<br>
Je to dobrá příležitost pro debatu o tom, jak "ošetřit" vstupy od uživatele (validace).
</div>



````python
kolekce = "60,64,67".split(",")
a, b, c = kolekce
play(a)
play(b)
play(c)
````

Musíte si však dát pozor, abyste rozdělovali řetězec správným separátorem do kolekce, nebo naopak řetězec správně zadávali. Separátorem mohou být i bílé znaky jako je nový řádek ```"\n"```, tabulátor ```"\t"``` nebo mezera ```" "```. Také se sepárator může skládat i z více znaků než z jednoho. Následující příkaz nerozdělí řetězec do kolekce, protože separátorem je mezera a nikoliv čárka.

````python
kolekce = "60 64 67".split(",")
a, b, c = kolekce
play(a)
play(b)
play(c)
````

Pokud je separátorem mezera, pak nemusíte udávat separující znak. Následující dva programy fungují stejně.

````python
a, b, c = "60 64 67".split(" ")
play(a)
play(b)
play(c)
````

````python
a, b, c = "60 64 67".split("")
play(a)
play(b)
play(c)
````

Samozřejmě příkaz split (konkrétně se jedná o metodu, čemuž porozumíte na konci našeho kurzu) lze volat i nad proměnnou, ve které se řetězec nachází. Jazyk Python nejprve vyvolá z proměnné hodnotu a následně zavolá metodu split.

````python
tony = "60, 64, 67"
a, b, c = tony.split(",")
play(a)
play(b)
play(c)
````

### Cvičení 2.3.5: Náhodné elementy

Napište program, který požádá uživatele o zadání 5 libovolných tónů (ve slyšitelném spektru: 25-80). Tóny budou odděleny vámi vybraným separátorem.<br>Rozdělte data získaná z výzvy podle zvoleného separátoru a přehrajte dané proměnné s náhodně velkou pauzou v rozmezí od 0 do 3 včetně, náhodnou stranovostí zvuku (pan) a náhodnou amplitudou (amp).

Najděme vhodnou funkci z modulu random [ZDE](https://docs.python.org/3/library/random.html)


<div style="color:#90EE90">

#### Tip pro vyučující:
V tomto cvičení žáci ručně vypisují play a sleep pro každý tón zvlášť. Je to záměrné! Máme u nich vyvolat pocit "tohle je nuda a zdlouhavé", abychom je v lekci věnované cyklům (Lekce 4) mohli nadšeně seznámit s cykly, které tuto práci udělají za ně na dvou řádcích.

Nezapomeňte žákům připomenout, že aby fungovalo random.uniform, musí být na začátku (nebo v buňce) import random. Bez toho by nastal NameError: name 'random' is not defined.
</div>



````python
tony = list(map(int, input("Zadej 5 tónů hodnoty 25-80 oddělených čárkou: ").split(",")))
A, B, C, D, E = tony
pauza = random.uniform(0,3)
strana = random.uniform(0,1)
hlasitost = random.random()
play(A, pan=strana, amp=hlasitost)
sleep(pauza)
pauza = random.uniform(0,3)
strana = random.uniform(0,1)
hlasitost = random.random()
play(B, pan=strana, amp=hlasitost)
sleep(pauza)
pauza = random.uniform(0,3)
strana = random.uniform(0,1)
hlasitost = random.random()
play(C, pan=strana, amp=hlasitost)
sleep(pauza)
pauza = random.uniform(0,3)
strana = random.uniform(0,1)
hlasitost = random.random()
play(D, pan=strana, amp=hlasitost)
sleep(pauza)
strana = random.uniform(0,1)
hlasitost = random.random()
play(E, pan=strana, amp=hlasitost)
````
