### Kapitola 4.1 Cyklus While

Nejjednodušším typem cyklu je cyklus while. Tento cyklus vykonává příkazy ve svém bloku do té doby, dokud je podmínka vyhodnocena jako pravdivá. V hudbě to často znamená: "Hraj tento rytmus, dokud tě nezastavím".

```python
opakovani = 0 #tempo v úderech za minutu

while opakovani < 5:
    play(60)
    sleep(0.5)
    opakovani += 1
```

V předchozím příkladu vidíte, jak se využívá cyklus while k postupné změně hodnoty proměnné opakovani. Dokud je splněna podmínka, že opakovani je nižší než 10, program opakuje úder na notu. Jakmile hodnota dosáhne limitu, cyklus skončí a hudba utichne. Proměnná opakovani zde funguje jako řídicí proměnná.

V každém průchodu cyklem se únava zvýší o 1. <br>Nezpomeňte dát příkaz sleep() s libolnou hodnotou, na vhodné místo v cyklu, jinak všechny zvuky zazní v podstatě v jeden okamžik.

Složené podmínky v rytmu

Jelikož podmínka za příkazem _while_ je logický výraz, může být i složená. To se hodí, když chceme, aby hudba hrála, jen pokud jsou splněny určité parametry.
Následující cyklus bude generovat náhodné tóny, dokud bude hlasitost (amp) v bezpečných mezích a zároveň budeme mít dostatek "energie".

```python
from random import randint

energie = 7
hlasitost = 0.1

while energie > 0 and hlasitost <= 1.5:
    play(randint(50, 80), amp = hlasitost) # Zahraje náhodnou notu s postupěn se měnící hlasitostí
    sleep(0.25)
    
    energie -= 1
    hlasitost += 0.2
```

V tomto případě musí být splněny obě podmínky. Jakmile umělec ztratí energii (dojde na nulu) nebo hudba překročí nastavenou hlasitost, cyklus se přeruší a nastane ticho.

### Kapitola 4.2: Cyklus For

V programování je cyklus for neuvěřitelně mocný nástroj. Zatímco while používáme spíše pro nekonečné smyčky cyklus for používáme pro přesně definované úseky (Víme kolikrát má cyklus proběhnout, ještě před spuštěním tohoto cyklu).

Zápis je mnohem čistší, protože se nemusíme starat o ruční zvyšování hodnoty řídící proměnné.<br>
Podívejme se níže na dva funkčně totožné programy. Jeden je zapsán pomocí _While_ cyklu, druhý pomocí _For_ cyklu.

```python
nota = 60
while nota <= 72:
    play(nota)
    sleep(0.25)
    nota += 1
```

```python
for nota in range(60, 72, 1):
    play(nota)
    sleep(0.25)
```

Místo ručního počítání řekneme počítači, odkud kam má "dojít" a jak velké kroky má dělat.
Můžeme si to představit následovně

```python
start = 0      # První nota
konec = 11      # Tady se zastavíme
krok = 1        # Posun (v hudbě např. o půltón)

for nota in range(start, konec, krok):
    play(nota)  # Zahraje tón
    sleep(0.5)
```

hodnota _konec_, v programu výše, není zahrnuta do cyklu. Jinými slovy cyklus je vykonán od čísla 0 do 10 s krokem 1.

## Kapitola 4.3: Předčasné ukončení běhu smyčky

Někdy chceme, aby cyklus skončil dříve, jakmile narazí na "zakázanou" notu nebo když dosáhne určité hudební hranice. K tomu slouží příkaz _break_.
<br>
V následujícím příkladu necháme program generovat vzestupnou řadu tónů, ale jakmile narazíme na tón, který je v naší stupnici "nepovolený", celou sekvenci okamžitě ukončíme.

```python
for nota in range(57, 85):
    if nota % 9 == 0:
        break  # Okamžitě vyskočí z cyklu ven    
    play(nota, release=0.1)
    sleep(0.2)

#když cyklus doběhne ozve se gong
use_synth(HOLLOW)
play(40, attack=0.01, release=8, amp=2)
```

Zamysleme se, na notě jaké hodnoty se cyklus zastavil?

Obdobně lze předčasně ukončit while cyklus. Následující program nechá uživatele opakovaně hádat jeden ze 4 tonů (60 až 63), Pokud se uživatel trefí, započte mu bod k vítězství a přehraje daný tón, pokud se netrefí, zahraje ton "chyby". Pokud uživatel bude mít 2 vítězství, cyklus předčasně končí.

```python
import random

pocet_vitezstvi = 0
celkovy_pocet_kol = 5
aktualni_kolo = 1


while aktualni_kolo <= celkovy_pocet_kol: 
    hadany_ton = random.randint(60, 63)
    zadany_ton = int(input(f"Kolo {aktualni_kolo}/{celkovy_pocet_kol}. Hádej tón (60 až 63): "))
    
    if zadany_ton == hadany_ton:
        pocet_vitezstvi += 1
        play(zadany_ton, release=0.5)
    
    if zadany_ton != hadany_ton:
        use_synth(SAW)
        play(37, release=0.2)
        play(38, release=0.2)

    if pocet_vitezstvi == 2:
        use_synth(PROPHET)
        play([60, 64, 67, 72], attack=0.1, release=2, amp=0.8)
        break
    
    aktualni_kolo += 1
    sleep(0.3)

sleep(0.1)
use_synth(SINE)
play(36, attack=0.5, release=4, amp=1.5)
```

## Kapitola 4.4: Zanořené cykly v hudební struktuře

V programování i hudbě můžeme vkládat jeden cyklus do druhého. Tato technika, nazývaná zanořování, je v Sonic Pi naprosto zásadní pro vytváření komplexních rytmických struktur, jako jsou melodické linky běžící přes několik taktů nebo vrstvení bicích smyček.

Představte si vnější cyklus jako počet opakování celého hudebního motivu a vnitřní cyklus jako jednotlivé tóny v rámci tohoto motivu. V kódu níže vidíte vnější cyklus s proměnnou _i_ a vnitřní cyklus s proměnnou _j_.

```python
for i in range(5):
    for j in range(3):
        play(60 + j) # Přehrává tóny, vždy o půlton vyšší, celkem i-krát
        sleep(0.25)
```

anořování můžete vrstvit téměř do nekonečna. Cykly mohou stát vedle sebe, nebo se do sebe vnořovat v mnoha úrovních, přičemž uvnitř nich lze používat podmínky

V následujícím příkladu použijeme zanořené cykly k vytvoření dynamické skladby. Vnější cyklus určuje počet opakování celého bloku, zatímco vnitřní cykly generují různé rytmické variace založené na matematických podmínkách (např. dělitelnost čísla tónu)

```python
for i in range(2):
    for j in range(8):
        for k in range(5):
            if k % 2 == 0:
                use_synth(PROPHET)
                play(70 + k, release=0.1)
        
        for l in range(10):
            if l % 2 != 0:
                sample(DRUM_SNARE_HARD)
           
        sleep(0.125)
```
